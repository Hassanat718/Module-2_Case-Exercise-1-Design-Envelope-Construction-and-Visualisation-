# **Implement the corrosion model in Python**

```python
def calculate_corrosion_rate(
    initial_thickness_mm,
    current_thickness_mm,
    elapsed_years
):
    if elapsed_years <= 0:
        raise ValueError("Elapsed time must be positive.")

    if initial_thickness_mm < current_thickness_mm:
        raise ValueError(
            "Current thickness exceeds initial thickness. "
            "Check the measurements."
        )

    return (
        initial_thickness_mm - current_thickness_mm
    ) / elapsed_years


def calculate_rul(
    current_thickness_mm,
    minimum_thickness_mm,
    corrosion_rate_mm_yr
):
    if current_thickness_mm < minimum_thickness_mm:
        raise ValueError(
            "Current thickness is below the minimum allowable thickness."
        )

    if corrosion_rate_mm_yr < 0:
        raise ValueError("Corrosion rate cannot be negative.")

    if corrosion_rate_mm_yr == 0:
        return float("inf")

    return (
        current_thickness_mm - minimum_thickness_mm
    ) / corrosion_rate_mm_yr
```

# **Run the model**

```python
initial_thickness = 12.7
current_thickness = 10.8
minimum_thickness = 8.5
elapsed_years = 10

corrosion_rate = calculate_corrosion_rate(
    initial_thickness,
    current_thickness,
    elapsed_years
)

rul = calculate_rul(
    current_thickness,
    minimum_thickness,
    corrosion_rate
)

print(f"Corrosion rate: {corrosion_rate:.3f} mm/year")
print(f"Remaining useful life: {rul:.2f} years")
```

# **Validate against a hand calculation**

```python
# Independent hand calculations
hand_corrosion_rate = (12.7 - 10.8) / 10
hand_rul = (10.8 - 8.5) / hand_corrosion_rate

# Automated validation
assert abs(corrosion_rate - hand_corrosion_rate) < 1e-9
assert abs(rul - hand_rul) < 1e-9

print("Corrosion model validation: PASSED")
print(f"Hand-calculated corrosion rate: {hand_corrosion_rate:.3f} mm/year")
print(f"Python corrosion rate:          {corrosion_rate:.3f} mm/year")
print(f"Hand-calculated RUL:             {hand_rul:.2f} years")
print(f"Python RUL:                      {rul:.2f} years")
```

# **Implement the hoop stress model**

```python
def calculate_hoop_stress(
    pressure_mpa,
    outside_diameter_mm,
    wall_thickness_mm
):
    if pressure_mpa < 0:
        raise ValueError("Pressure cannot be negative.")

    if outside_diameter_mm <= 0:
        raise ValueError("Diameter must be positive.")

    if wall_thickness_mm <= 0:
        raise ValueError("Wall thickness must be positive.")

    if 2 * wall_thickness_mm >= outside_diameter_mm:
        raise ValueError(
            "The thin-wall approximation may not be appropriate."
        )

    hoop_stress_mpa = (
        pressure_mpa * outside_diameter_mm
    ) / (2 * wall_thickness_mm)

    return hoop_stress_mpa
```

# **Calculate hoop stress**

```python
pressure = 10.0
outside_diameter = 273.1
wall_thickness = 12.7

hoop_stress = calculate_hoop_stress(
    pressure,
    outside_diameter,
    wall_thickness
)

print(f"Estimated hoop stress: {hoop_stress:.2f} MPa")
```

# **Validate against a hand calculation**

```python
# Independent hand calculation
hand_hoop_stress = (
    10.0 * 273.1
) / (
    2 * 12.7
)

# Automated validation
assert abs(hoop_stress - hand_hoop_stress) < 1e-9

print("Hoop stress model validation: PASSED")
print(f"Hand-calculated hoop stress: {hand_hoop_stress:.2f} MPa")
print(f"Python hoop stress:          {hoop_stress:.2f} MPa")
```

# **QUESTION 4**
Create a risk matrix plot for the subsea Christmas tree FMECA

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Confirm that the required columns exist
required_columns = [
    "Failure_Mode_ID",
    "Severity",
    "Occurrence",
    "Detection",
    "RPN"
]

assert all(col in fmeca.columns for col in required_columns), (
    "The FMECA DataFrame is missing required columns."
)

# Recalculate RPN to ensure consistency
fmeca["RPN"] = (
    fmeca["Severity"]
    * fmeca["Occurrence"]
    * fmeca["Detection"]
)

print(f"Number of failure modes: {len(fmeca)}")
print("Risk matrix data prepared successfully.")
```

# **Plot the risk matrixA**

```python
fig, ax = plt.subplots(figsize=(13, 9))

# Illustrative risk regions based on severity × occurrence
# The RPN annotations will use the full S × O × D calculation.
risk_regions = [
    (1, 1, 4, 4, "Low"),
    (1, 5, 4, 6, "Moderate"),
    (1, 7, 4, 10, "High"),
    (5, 1, 6, 4, "Moderate"),
    (5, 5, 6, 6, "High"),
    (5, 7, 6, 10, "Very High"),
    (7, 1, 10, 4, "High"),
    (7, 5, 10, 10, "Very High")
]

region_styles = {
    "Low": "lightgreen",
    "Moderate": "khaki",
    "High": "orange",
    "Very High": "salmon"
}

for x_start, y_start, x_end, y_end, label in risk_regions:
    ax.fill_between(
        [x_start - 0.5, x_end + 0.5],
        y_start - 0.5,
        y_end + 0.5,
        color=region_styles[label],
        alpha=0.22,
        zorder=0
    )

# Plot each failure mode
ax.scatter(
    fmeca["Occurrence"],
    fmeca["Severity"],
    s=100,
    edgecolors="black",
    zorder=3
)

# Annotate each point with its failure mode ID and RPN
for _, row in fmeca.iterrows():
    ax.annotate(
        f'{row["Failure_Mode_ID"]}\nRPN={row["RPN"]}',
        (row["Occurrence"], row["Severity"]),
        xytext=(6, 5),
        textcoords="offset points",
        fontsize=8,
        zorder=4
    )

ax.set_title(
    "Subsea Christmas Tree FMECA Risk Matrix",
    fontsize=15
)
ax.set_xlabel("Occurrence Rating (1–10)", fontsize=12)
ax.set_ylabel("Severity Rating (1–10)", fontsize=12)

ax.set_xticks(range(1, 11))
ax.set_yticks(range(1, 11))
ax.set_xlim(0.5, 10.5)
ax.set_ylim(0.5, 10.5)

ax.grid(True, linestyle="--", alpha=0.35)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```
